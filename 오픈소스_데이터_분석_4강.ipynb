```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

# 계절별 순서
season_order = ["봄", "여름", "가을", "겨울"]
# 제목
plt_title = "계절별 배추 평균 가격"
## 데이터 수집 url 및 api key 설정
url = "https://apis.data.go.kr/B552845/perDay/price"
# api key
api_key = "b5c6987076c5e9a3b931b248881047aeb7af6bd5269e501306db93546fb6418c"
# 품종코드
prod_codes = {
    "01": "봄",
    "02": "여름",
    "03": "가을",
    "06": "겨울"
}
# 시작월
year_from = 2015
# 종료월
year_to = 2025
# 응답 JSON items
resp_items = []

for year in range(year_from, year_to):

    for prod_code, prod_name in prod_codes.items():
        params = {
            'serviceKey':api_key,
            'pageNo':'1',
            'numOfRows':'200',
            'returnType':'JSON',
            'cond[exmn_ymd::GTE]':f'{year}0101',
            'cond[exmn_ymd::LTE]':f'{year}1231',
            'cond[se_cd::EQ]':'02',
            'cond[ctgry_cd::EQ]':'200',
            'cond[item_cd::EQ]':'211',
            'cond[grd_cd::EQ]':'04',
            'cond[sgg_cd::EQ]':'1101',
            'cond[mrkt_cd::EQ]':'0110211',
            'cond[vrty_cd::EQ]':prod_code
        }
        # 데이터 수집
        response = requests.get(url, params=params)

        # 호출 성공/실패 출력
        resp_json = response.json()

        # API 응답 데이터 확인
        items = resp_json.get("response", {}).get("body", {}).get("items", {}).get("item", {})

        # 응답 items 리스트를 응답 items 리스트에 추가 처리
        for item in items:
            item["year"] = year
            item["season"] = prod_name
            resp_items.append(item)


# 수집한 JSON 데이터를 DataFrame으로 변환
df = pd.DataFrame(resp_items)

# 가격을 숫자형으로 변환
df["exmn_dd_prc"] = pd.to_numeric(df["exmn_dd_prc"])

# 계절별 순서 대로 정렬
df["season"] = pd.Categorical(df["season"], categories=season_order, ordered=True)

# 계절별 배추가격을 group by 평균
season_avg = df.groupby("season", observed=True)["exmn_dd_prc"].mean()

# 막대 그래프
plt.figure(figsize=(8, 5))

bars = plt.bar(
    season_avg.index,
    season_avg.values
)

plt.title("계절별 배추 평균 가격")
plt.xlabel("계절")
plt.ylabel("평균 가격")

# 각 막대에 구체적인 수치 표시
for bar, value in zip(bars, season_avg.values):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:,.2f}",
        ha="center",
        va="bottom"
    )

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.show()
```
