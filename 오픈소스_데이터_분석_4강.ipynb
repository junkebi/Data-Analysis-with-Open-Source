```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

# 제목 - "연도별 배추 평균 가격 변화 - (학번뒤4자리)"
plt_title = "연도별 배추 평균 가격 변화 - 2338"
url = "https://apis.data.go.kr/B552845/perDay/price"
api_key = "b5c6987076c5e9a3b931b248881047aeb7af6bd5269e501306db93546fb6418c"
prod_codes = {"01": "봄배추", "02": "여름배추(고랭지)", "03": "가을배추", "06": "월동배추"}
resp_items = []

for year in range(2015, 2025):
    for prod_code, prod_name in prod_codes.items():
        params = {
            'serviceKey':api_key,    'pageNo':'1',                        'numOfRows':'200',
            'returnType':'JSON',     'cond[exmn_ymd::GTE]':f'{year}0101', 'cond[exmn_ymd::LTE]':f'{year}1231',
            'cond[se_cd::EQ]':'02',  'cond[ctgry_cd::EQ]':'200',          'cond[item_cd::EQ]':'211',
            'cond[grd_cd::EQ]':'04', 'cond[sgg_cd::EQ]':'1101',           'cond[mrkt_cd::EQ]':'0110211',
            'cond[vrty_cd::EQ]':prod_code
        }
        # 데이터 수집
        response = requests.get(url, params=params)
        resp_json = response.json()
        items = resp_json.get("response", {}).get("body", {}).get("items", {}).get("item", {})
        resp_items.extend(items)

# 수집한 JSON 데이터를 DataFrame으로 변환
df = pd.DataFrame(resp_items)
# 조사일자를 날짜형으로 변환
df["exmn_ymd"] = pd.to_datetime(df["exmn_ymd"].astype(str))
# 연도 추출
df["year"] = df["exmn_ymd"].dt.year

# 가격 데이터를 숫자형으로 변환
df["exmn_dd_prc"] = pd.to_numeric(df["exmn_dd_prc"])
# 연도별 평균 가격 계산
year_avg = df.groupby("year")["exmn_dd_prc"].mean()

# 선 그래프
plt.figure(figsize=(10, 5))

plt.plot(
    year_avg.index,
    year_avg.values,
    marker="o",
    linewidth=2
)

plt.title(plt_title)  # 학번 뒤 4자리 수정
plt.xlabel("연도")
plt.ylabel("평균 가격")
plt.xticks(year_avg.index)
plt.grid(True, alpha=0.3)

plt.tight_layout()

# 그래프 저장
plt.savefig(f'{plt_title}.png', dpi=300)

plt.show()
```
