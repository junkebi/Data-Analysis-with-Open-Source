```python
import requests
import pandas as pd

## 데이터 수집 url 및 api key 설정
url = "https://apis.data.go.kr/B552845/perDay/price"
# api key
api_key = "b5c6987076c5e9a3b931b248881047aeb7af6bd5269e501306db93546fb6418c"
# 품종코드
prod_codes = {
    "01": "봄배추",
    "02": "여름배추(고랭지)",
    "03": "가을배추",
    "06": "월동배추"
}
# 시작월
year_from = 2015
# 종료월
year_to = 2025
# 응답 JSON items
resp_items = []

for year in range(year_from, year_to):

    for prod_code, prod_name in prod_codes.items():
        params = {
            'serviceKey':api_key,
            'pageNo':'1',
            'numOfRows':'200',
            'returnType':'JSON',
            'cond[exmn_ymd::GTE]':f'{year}0101',
            'cond[exmn_ymd::LTE]':f'{year}1231',
            'cond[se_cd::EQ]':'02',
            'cond[ctgry_cd::EQ]':'200',
            'cond[item_cd::EQ]':'211',
            'cond[grd_cd::EQ]':'04',
            'cond[sgg_cd::EQ]':'1101',
            'cond[mrkt_cd::EQ]':'0110211',
            'cond[vrty_cd::EQ]':prod_code
        }
        # 데이터 수집
        response = requests.get(url, params=params)

        # 호출 성공/실패 출력
        resp_json = response.json()

        # API 응답 데이터 확인
        items = resp_json.get("response", {}).get("body", {}).get("items", {}).get("item", {})

        # 응답 items 리스트를 응답 items 리스트에 추가 처리
        for item in items:
            item["year"] = year
            item["variety"] = prod_name
            item["variety_code"] = prod_code
            resp_items.append(item)

    print("-" * 60)

print("=" * 60)
# 수집한 JSON 데이터를 DataFrame으로 변환
df = pd.DataFrame(resp_items)

print("===== DataFrame 기본 정보 =====")
print("행(row) 수 :", df.shape[0])
print("열(column) 수 :", df.shape[1])

print("\n===== 컬럼명 =====")
print(df.columns.tolist())

print("\n===== 데이터 타입 =====")
print(df.dtypes)

print("\n===== 데이터 앞부분 =====")
print(df.head())

print("\n===== 기술통계 =====")
print(df.describe(include="all"))
print("=" * 60)

```

```python
import requests
import pandas as pd

## 데이터 수집 url 및 api key 설정
url = "https://apis.data.go.kr/B552845/perDay/price"
# api key
api_key = "b5c6987076c5e9a3b931b248881047aeb7af6bd5269e501306db93546fb6418c"
# 품종코드
prod_codes = {
    "01": "봄배추",
    "02": "여름배추(고랭지)",
    "03": "가을배추",
    "06": "월동배추"
}
# 시작월
year_from = 2015
# 종료월
year_to = 2025
# 응답 JSON items
resp_items = []

for year in range(year_from, year_to):

    for prod_code, prod_name in prod_codes.items():
        params = {
            'serviceKey':api_key,
            'pageNo':'1',
            'numOfRows':'200',
            'returnType':'JSON',
            'cond[exmn_ymd::GTE]':f'{year}0101',
            'cond[exmn_ymd::LTE]':f'{year}1231',
            'cond[se_cd::EQ]':'02',
            'cond[ctgry_cd::EQ]':'200',
            'cond[item_cd::EQ]':'211',
            'cond[grd_cd::EQ]':'04',
            'cond[sgg_cd::EQ]':'1101',
            'cond[mrkt_cd::EQ]':'0110211',
            'cond[vrty_cd::EQ]':prod_code
        }
        # 데이터 수집
        response = requests.get(url, params=params)

        # 호출 성공/실패 출력
        resp_json = response.json()

        # API 응답 데이터 확인
        items = resp_json.get("response", {}).get("body", {}).get("items", {}).get("item", {})

        # 응답 items 리스트를 응답 items 리스트에 추가 처리
        for item in items:
            item["year"] = year
            item["variety"] = prod_name
            item["variety_code"] = prod_code
            resp_items.append(item)

    print("-" * 60)

print("=" * 60)
# 수집한 JSON 데이터를 DataFrame으로 변환
df = pd.DataFrame(resp_items)

# 조사일자를 날짜형으로 변환
df["exmn_ymd"] = pd.to_datetime(
    df["exmn_ymd"].astype(str)
)

# 연도 추출
df["year"] = df["exmn_ymd"].dt.year


# 계절 구분 함수
def get_season(month):
    match month:
        case 3 | 4 | 5:
            return "봄"
        case 6 | 7 | 8:
            return "여름"
        case 9 | 10 | 11:
            return "가을"
        case 12 | 1 | 2:
            return "겨울"
        case _:
            return "알 수 없음"
    """
    if month in [3, 4, 5]:
        return "봄"

    elif month in [6, 7, 8]:
        return "여름"

    elif month in [9, 10, 11]:
        return "가을"

    else:
        return "겨울"
    """

# 계절 컬럼 생성
#df["season"] = df["exmn_ymd"].dt.month.apply(get_season)

season_list = []

for month in df["exmn_ymd"].dt.month:
    season_list.append(get_season(month))

df["season"] = season_list


# 결과 확인
print("===== 연도/계절 컬럼 추가 결과 =====")
print(df[["exmn_ymd", "year", "season"]].head(20))
```
